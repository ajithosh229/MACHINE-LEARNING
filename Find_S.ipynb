```python
def find_s_algorithm(filename):
  with open(filename,'r') as file:
    reader=csv.reader(file)
    data = list(reader)

  headers = data[0]

  training_data = data[1:]

  target_index = len(headers)-1

  attributes = headers[1:target_index]

  hypothesis = ['0'] * len(attributes)

  print("Attributes:",attributes)
  print("\nFIND-S Algorithm Steps")
  print("-" * 60)

  for row in training_data:
    example = row[1:target_index]
    target = row[target_index]

    if target.lower() == "yes":

      print("\nTraining Example:",row[0])
      print("Positive Example:",example)

      if hypothesis == ['0'] * len(attributes):
        hypothesis = example.copy()

      else:
        for i in range(len(hypothesis)):
          if hypothesis[i] != example[i]:
            hypothesis[i] = '?'
      print("Hypothesis:",hypothesis)
    else:
      print("\nTraining Example:",row[0])
      print("Negative Example - Ignored")
  return hypothesis
```

```python
final_hypothesis = find_s_algorithm("/content/drive/MyDrive/enjoysport.csv")
```

```python
print("\n" + "=" * 60)
print("Final Most Specific Hypothesis:")
print(final_hypothesis)
print("=" * 60)
```
